# Crédito de custeio rural e produção agrícola no Brasil

**Uma análise exploratória do Crédito Rural e da Produção Agrícola das culturas de soja soja, milho e cana-de-açúcar entre 2016 e 2025**

## Resumo

Este trabalho integra dados municipais de crédito de custeio do Banco Central do Brasil (BCB) e indicadores da Produção Agrícola Municipal (PAM) do IBGE. O objetivo é descrever a evolução e a distribuição territorial do crédito e comparar seu volume com a escala e a produtividade agrícola. A base integrada reúne 164.775 observações de município, cultura e ano, das quais 52.652 atendem aos critérios de validação. A soja liderou o volume nominal anual no período. No Top 10% dos municípios, o crédito ficou mais concentrado que a produção para a cana-de-açúcar. Para milho e soja, a produção ficou mais concentrada. Entre os municípios que cultivaram soja em 2025, as medianas de produtividade oscilaram entre os quartis de crédito por hectare. Os resultados são descritivos e dependem da cobertura de cada fonte.

**Palavras-chave** — crédito rural, custeio agrícola, produtividade, produção agrícola, crédito rural.

## 1. Introdução e proposta do trabalho

O crédito de custeio rural financia despesas do ciclo agrícola, como a compra de sementes, fertilizantes e outros insumos. O projeto investiga como esse financiamento se distribuiu pelo Brasil entre 2016 e 2025 e como os valores contratados se relacionam com a área plantada, a quantidade produzida e a produtividade de soja, milho e cana-de-açúcar.

A pergunta central é se o crédito acompanha a escala da atividade agrícola ou se algumas regiões e municípios recebem uma parcela do financiamento desproporcional à sua participação na área ou na produção. Uma região pode receber muito crédito porque concentra grande parte da área plantada. Por isso, o estudo compara tanto o volume contratado quanto medidas relativas, como a participação de cada região no crédito nacional e o crédito por hectare ou por tonelada. As três culturas são analisadas separadamente porque sua produção varia no território nacional e desejava-se entendê-las individualmente para analizar como o crédito rural e a produção agrícola de cada cultura está no Brasiç. Neste trabalho, uma participação desproporcional significa apenas que as parcelas de crédito e de área ou produção são diferentes. O estudo não tenta estabelecer qual seria a distribuição ideal do financiamento.

A análise parte da evolução nacional ao longo do tempo, passa pela distribuição entre regiões e estados e chega à comparação entre municípios. Nessa última etapa, o trabalho busca entender quanto o crédito varia entre municípios com áreas semelhantes, se os grupos com maior crédito por hectare apresentam maior produtividade e quais municípios se afastam do padrão do próprio estado.

O estudo usa dados agregados por município e descreve padrões observados. Uma associação entre crédito e produtividade pode refletir diferenças locais de clima, solo, tecnologia ou acesso ao financiamento. Os dados também não permitem acompanhar o crédito e a produção do mesmo estabelecimento rural.

## 2. Dados e metodologia

A base de crédito veio do Sistema de Operações do Crédito Rural e do Proagro (SICOR), do Banco Central do Brasil. O projeto usou o valor contratado e a área financiada, identificados por município, cultura e mês de emissão. Os valores mensais foram somados para construir a base anual. Os dados são de custeio agrícola e representam uma parte do crédito rural.

A base agrícola veio da Produção Agrícola Municipal (PAM), do IBGE. Foram usados a área plantada, em hectares, a quantidade produzida, em toneladas, e o rendimento médio oficial, originalmente em quilogramas por hectare. Os dados foram consultados por município, estado ou região, conforme a pergunta. Os mapas usam os limites municipais e estaduais do IBGE de 2020.

Para aproximar as duas fontes, o projeto padronizou os nomes das culturas e os identificadores de município. Uma observação corresponde a um município, uma cultura e um ano. A base integrada indica se há dados nas duas fontes ou em apenas uma delas. Na PAM, os valores registrados como zero foram mantidos e os valores indisponíveis, omitidos ou sem aplicação ficaram ausentes.

Para as comparações que exigiam crédito e área na mesma observação, foram mantidos os registros presentes nas duas fontes, com área plantada positiva e valor de crédito válido igual ou superior a zero. A área usada no cálculo do crédito por hectare é a área plantada registrada pelo IBGE. A área financiada é uma medida do Banco Central. As contagens municipais de contratos estavam ausentes nos dados analisados.

## 3. Resultados e leitura dos sete gráficos

Cada subseção explica a pergunta que motivou o gráfico e como interpretar seus eixos, barras, pontos ou cores.

### 3.1. Evolução do crédito e dos contratos

O primeiro painel do gráfico 1 mostra, no eixo horizontal, os meses de emissão e, no vertical, o valor contratado em reais correntes. Cada linha representa uma cultura. Picos indicam meses de maior contratação e a distância entre linhas permite comparar os valores no mesmo mês. Seu segundo painel, rotulado “Quantidade de Contratos”, usa a contagem de linhas da resposta agrupadas por mês e cultura. A evolução do número de contratos fica em aberto, pois a unidade de uma linha é um registro da API. Os valores anuais citados a seguir somam os meses de cada cultura e são expressos em bilhões de reais (R$) nominais.

### 3.2. Distribuição por região e UF

O gráfico 2 compara a participação de cada região e UF no total nacional de crédito e de área plantada, para os anos comuns de cada cultura. Cada indicador tem seu próprio denominador. Os oito estados com maior área aparecem individualmente e os demais integram “Outros”. Uma região com uma parcela do crédito superior à da área tem maior peso relativo no financiamento. Para ler o gráfico, compare as duas barras da mesma região ou UF. Uma diferença de dez pontos percentuais significa, por exemplo, que uma região concentrou 30% do crédito e 20% da área nacional daquela cultura. Os percentuais citados a seguir usam os totais municipais da base integrada.

### 3.3. Concentração municipal

O gráfico 3 soma os anos comuns por município e ordena separadamente os totais de crédito e de produção. O número de municípios do Top 1%, 5% e 10% é arredondado para cima. Os grupos são cumulativos e podem conter municípios diferentes em cada ranking. O indicador ausente recebe zero para o cálculo da participação no ranking. A ausência continua registrada na base integrada. Para avaliar a concentração, compare as barras de crédito e produção dentro da mesma faixa. “Demais municípios” reúne os que ficam fora do Top 10% em cada ranking.

### 3.4. Crédito entre municípios com áreas semelhantes

No gráfico 4, cada ponto é um município da cultura e do ano selecionados. O eixo horizontal mostra a área plantada, o vertical mostra o crédito nominal e as cores indicam a região. A distância vertical entre pontos com áreas próximas mostra quanto varia o crédito em municípios de escala semelhante. O filtro de UF permite restringir a comparação.

Para resumir essa variação, a integração começa na menor área de cada cultura e ano e multiplica o limite inferior de cada faixa por 1,25 para obter o seguinte. Cada faixa inclui o limite inferior e exclui o superior. O resumo informa a mediana da área e as medianas, quartis e intervalos interquartis do crédito total e do crédito por hectare. O intervalo interquartil equivale a Q3 menos Q1 e mede a amplitude da metade central da distribuição.

### 3.5. Crédito por hectare e por tonelada

O gráfico 5 calcula R$/ha dividindo a soma do crédito pela soma da área plantada. Para R$/t, divide a soma do crédito pela produção total dos municípios com produção finita e positiva. Em cada razão, o numerador e o denominador incluem os mesmos municípios. A linha nacional segue a mesma regra. A razão de somas dá maior peso aos municípios com denominadores maiores e pode diferir da mediana das razões municipais.

### 3.6. Crédito por hectare e produtividade

O gráfico 6 exige pelo menos 50 ha plantados, crédito e rendimento positivos. O rendimento oficial é dividido por mil para ser expresso em t/ha. Em cada cultura, ano e recorte com pelo menos oito observações, os municípios são ordenados em quatro grupos de tamanhos semelhantes para cada indicador. Empates são resolvidos pela ordem dos registros. Os dois mapas mostram, separadamente, em que quartil cada município ficou para crédito por hectare e para produtividade. Os tons mais escuros representam os grupos de valores mais altos. No boxplot, cada caixa representa um quartil de crédito, a linha dentro da caixa indica a mediana da produtividade e a altura da caixa mostra a dispersão da metade central. Os pontos extremos ficam ocultos no boxplot. Registros sem polígono na malha de 2020 entram nas estatísticas e ficam fora dos mapas.

### 3.7. Municípios atípicos no padrão estadual

O gráfico 7 usa os mesmos filtros do gráfico 6 e exige pelo menos dez municípios elegíveis na UF para a cultura e o ano. Em cada UF, calcula os limites Q1 − 1,5 × IQR e Q3 + 1,5 × IQR. Aplica a regra ao logaritmo de base 10 de R$/ha e à produtividade em t/ha. Um município é sinalizado quando ultrapassa um limite e o IQR do indicador é positivo. A seleção geográfica muda a visualização, e o padrão de comparação continua sendo o da UF inteira.

O mapa usa azul para casos atípicos de crédito por hectare, verde para produtividade e vermelho para ambos. O cinza distingue municípios típicos de áreas sem dados ou excluídas, conforme a legenda. Na visualização de uma UF, até seis casos são identificados pelo nome, priorizando os que mais se distanciam da mediana estadual na escala padronizada.

## 4. Discussão e conclusão

### 4.1. Discussão

O trabalho buscou entender como o crédito de custeio evoluiu e se sua distribuição acompanhou a atividade agrícola. O primeiro gráfico mostra que o valor contratado variou ao longo do tempo e que as culturas tiveram trajetórias diferentes. A pergunta sobre a evolução do número de contratos ficou em aberto, pois o segundo painel apresenta a contagem de registros da base.

Nos gráficos de distribuição e concentração, o objetivo era verificar se o crédito acompanhava a área plantada e a produção. As barras de crédito e de área têm alturas diferentes em parte dos recortes, indicando que a participação de uma região ou estado no financiamento pode ficar acima ou abaixo da sua participação na área plantada. Na comparação municipal, a relação entre a concentração do crédito e a da produção depende da cultura e do grupo do ranking escolhido. Assim, a correspondência que se queria investigar foi observada apenas em parte.

A dispersão entre área plantada e crédito mostra que municípios com áreas próximas podem contratar valores diferentes. A comparação entre estados acrescenta outra perspectiva ao relacionar o valor contratado com a área plantada ou com o volume produzido. Cada painel ajuda a entender como a posição dos estados muda conforme a medida usada para comparar o crédito e como o crédito é distribuído em território nacional, como, por exemplo, perceber que estados como o Alagoas têm um recebimento de crédito muito acima da média nacional.

Na análise de produtividade, a pergunta era se os grupos com mais crédito por hectare também apresentavam maior rendimento. A leitura dos boxplots depende da cultura, do ano e do recorte selecionado. As caixas permitem observar se as medianas sobem de um grupo para outro e quanto as distribuições se sobrepõem. O último mapa complementa essa leitura ao mostrar quais municípios se destacam em crédito por hectare, em produtividade ou em ambos, em relação ao padrão do próprio estado.

### 4.2. Conclusão

Os gráficos mostram que o crédito acompanhou apenas em parte a distribuição da atividade agrícola. Regiões com grande participação na área plantada podem ter um peso diferente no financiamento, e municípios com áreas semelhantes podem contratar valores distintos. A análise atingiu o objetivo de descrever essas diferenças e mostrar como elas variam entre culturas e recortes.

A comparação por quartis permite examinar a relação entre crédito por hectare e produtividade no recorte escolhido. Com os dados agregados por município, o trabalho identifica diferenças entre os grupos observados e casos que se afastam do padrão estadual. Estimar o efeito do crédito sobre o rendimento exigiria outro desenho de análise.
